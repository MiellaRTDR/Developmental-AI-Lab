## Define the tutle environment class

You don't need to worry about the code of the ColabTurtleEnvironment below.

Just know that this environment:
* interprets the agent's actions as follows  `0`: move forward, `1`: turn left, `2`: turn right.
* returns outcome `1` when the turtle bumps into the border of the window, and `0` otherwise.

```python
# @title Initialize the turtle environment

BORDER_WIDTH = 20

class ColabTurtleEnvironment:

    def __init__(self):
        """ Creating the Turtle window """
        bgcolor("lightGray")
        penup()
        goto(window_width() / 2, window_height()/2)
        face(0)
        pendown()
        color("green")

    def outcome(self, action):
        """ Enacting an action and returning the outcome """
        _outcome = 0
        for i in range(10):
            # _outcome = 0
            if action == 0:
                # move forward
                forward(10)
            elif action == 1:
                # rotate left
                left(4)
                forward(2)
            elif action == 2:
                # rotate right
                right(4)
                forward(2)

            # Bump on screen edge and return outcome 1
            if xcor() < BORDER_WIDTH:
                goto(BORDER_WIDTH, ycor())
                _outcome = 1
            if xcor() > window_width() - BORDER_WIDTH:
                goto(window_width() - BORDER_WIDTH, ycor())
                _outcome = 1
            if ycor() < BORDER_WIDTH:
                goto(xcor(), BORDER_WIDTH)
                _outcome = 1
            if ycor() > window_height() - BORDER_WIDTH:
                goto(xcor(), window_height() -BORDER_WIDTH)
                _outcome = 1

            # Change color
            if _outcome == 0:
                color("green")
            else:
                # Finit l'interaction
                color("red")
                # if action == 0:
                #     break
                if action == 1:
                    for j in range(10):
                        left(4)
                elif action == 2:
                    for j in range(10):
                        right(4)
                break

        return _outcome
```

## Define the valence of interactions

```python
valences = [[1, -1],
            [-1, -1],
            [-1, -1]]
```

The valence table specifies the valence of each interaction. An interaction is a tuple (action, outcome):

|| 0 Not bump | 1 Bump|
|---|---|---|
| 0 Forward| 1 | -1 |
| 1 Left | -1 | -1 |
| 2 Right| -1 | -1 |

## Instantiate the agent

```python
a = Agent(valences)
```

## Run the simulation

```python
# @title Run the simulation

initializeTurtle()

# Parameterize the rendering
bgcolor("lightGray")
penup()
goto(window_width() / 2, window_height()/2)
face(0)
pendown()
color("green")
speed(10)

e = ColabTurtleEnvironment()

outcome = 0
for i in range(10):
    action = a.action(outcome)
    outcome = e.outcome(action)
```

## Create Agent3 by overriding the class Agent or your previous class Agent2

```python
import random
class Agent3:
    def __init__(self, _valences):
        """ Creating our agent """
        self._valences = _valences
        self._action = None
        self._predicted_outcome = None
        self._positive_action = None
        self._known_outcomes = {}
        self.cpt = 0

    def action(self, _outcome):
        """ tracing the previous cycle """
        if self._action is not None:
            print(f"Action: {self._action}, Prediction: {self._predicted_outcome}, Outcome: {_outcome}, "
                  f"Prediction: {self._predicted_outcome == _outcome}, Valence: {self._valences[self._action][_outcome]}")

        # outcome de l'action précédente
        if self._action is not None:
            self._known_outcomes[self._action] = _outcome

        # regarde si la valence est positive
        if self._action is not None and self._valences[self._action][_outcome] > 0:
            self._positive_action = self._action
            self.cpt += 1
        else:
            self.cpt = 0

        actions = [0, 1, 2]

        # calcul de la prochaine action
        if self._positive_action is None:
            # pas d'action positive explorée
            self._action = random.choice(actions)
        elif self._action is not None and self._valences[self._action][_outcome] < 0:
            # l'action précédente a produit un résultat négatif, on change
            self._action = random.choice([a for a in actions if a != self._action])
            self.cpt = 0
        elif self.cpt < 4:
            # préfère les actions positives
            self._action = self._positive_action
        else:
            # essayer d'autres actions par ennui (tourner)
            self._action = random.choice([1, 2])
            self.cpt = 0

        # anticipation
        self._predicted_outcome = self._known_outcomes.get(self._action, 0)

        return self._action
```
